```python
from pathlib import Path
import os
```

## Preparation
Next we need to define a `WORKING_DIR` where AlphaFold finds our input data and a `RESULTS_DIR` where the output files are written to. You can see these files in the file browser on the left. 


```python
ALPHAFOLD_WORKING_DIR = Path.cwd() / "alphafold2"
ALPHAFOLD_RESULTS_DIR = ALPHAFOLD_WORKING_DIR / "output"
```

## Input preparation

For the structure prediction we need the sequence in fasta format. Feel free to look at our example to adapt your input. 

Here we take human insulin from UniProt: https://www.uniprot.org/uniprotkb/P01308/entry

```python
INPUT_FASTA = ALPHAFOLD_WORKING_DIR / "insulin.fasta"
```

## Input File

Now we can use these directories to run  the structure prediction.
Execute the next cell to write the run file that controls the execution. You don't need to worry about the parameters too much. They are prepared for you. Only change them if you know what you're doing. Note that larger prediction runs might require different settings in the start of the jupyter session!

```python
ALPHAFOLD_RUN_PATH = ALPHAFOLD_WORKING_DIR / "run.sh"

run_file = f"""
#!/bin/bash
# AlphaFold 2 

# load software module 
module load bio/alphafold/2.3.2

# start program
run_alphafold.sh -d $ALPHAFOLD_DATABASES \\
    -o {ALPHAFOLD_RESULTS_DIR}  \\
    -f {INPUT_FASTA}  \\
    -t 2020-05-14 \\
    -c full_dbs \\
    -m multimer
    
"""

with open(ALPHAFOLD_RUN_PATH, "w") as file:
    file.write(run_file)
```

## Run the Diffusion

Execute the cell below to start the structure prediction. Good luck! This will take about 20 minutes.


```python
os.system(f'echo "Running file {ALPHAFOLD_RUN_PATH}"')
os.system(f"bash {ALPHAFOLD_RUN_PATH}")
```
